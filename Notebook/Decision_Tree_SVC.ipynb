# Diabetes Prediction using Decision Tree and SVC

This notebook demonstrates diabetes prediction using two machine learning algorithms:
- **Decision Tree Classifier**
- **Support Vector Classifier (SVC)**

We'll compare their performance and determine which model works best for diabetes prediction.

```python
# Import Required Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.preprocessing import StandardScaler
import warnings
warnings.filterwarnings('ignore')
```

```python
# Load the Diabetes Dataset
# You can use the Pima Indians Diabetes Database or any diabetes dataset
# For this example, we'll use sklearn's diabetes dataset or load from file

# Option 1: Load from sklearn (for demonstration)
from sklearn.datasets import load_diabetes

# Option 2: Load from CSV file (uncomment if you have a diabetes.csv file)
# df = pd.read_csv('../Dataset/diabetes.csv')

# For this demo, let's create a sample dataset or use a publicly available one
# We'll use the Pima Indians Diabetes dataset structure
np.random.seed(42)

# Create sample data with features similar to diabetes prediction
n_samples = 1000
data = {
    'Pregnancies': np.random.randint(0, 17, n_samples),
    'Glucose': np.random.normal(120, 30, n_samples),
    'BloodPressure': np.random.normal(69, 19, n_samples),
    'SkinThickness': np.random.normal(20, 15, n_samples),
    'Insulin': np.random.normal(80, 115, n_samples),
    'BMI': np.random.normal(32, 7, n_samples),
    'DiabetesPedigreeFunction': np.random.uniform(0.1, 2.5, n_samples),
    'Age': np.random.randint(21, 81, n_samples)
}

# Create target variable (0: No diabetes, 1: Diabetes)
# Make it somewhat realistic based on glucose and BMI
diabetes_prob = (data['Glucose'] > 126) | (data['BMI'] > 30)
data['Outcome'] = np.where(diabetes_prob & (np.random.random(n_samples) > 0.3), 1, 0)

df = pd.DataFrame(data)

print("Dataset shape:", df.shape)
print("\nFirst 5 rows:")
df.head()
```
